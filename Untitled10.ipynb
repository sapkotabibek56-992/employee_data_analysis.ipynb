6. Data Cleaning

```python
# Remove duplicate rows
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
# 6.2 STANDARDIZE GENDER
df["Gender"] = (
    df["Gender"]
    .replace({
        "female": "Female",
        "F": "Female",
        "male": "Male",
        "M": "Male"
    })
)

display(df["Gender"].value_counts(dropna=False))
```

```python
# 6.3 STANDARDIZE DEPARTMENT
df["Department"] = (
    df["Department"]
    .replace({
        "it": "IT",
        "Hr": "HR",
        "Human Resources": "HR"
    })
)

print(df["Department"].value_counts(dropna=False))
```

```python
# 6.4 STANDARDIZE OVERTIME
df["Overtime"] = (
    df["Overtime"]
    .str.strip()
    .str.lower()
    .map({
        "yes": "Yes",
        "no": "No"
    })
)

print(df["Overtime"].value_counts(dropna=False))
```

```python
# 6.5 CONVERT JOINDATE
df["JoinDate"] = pd.to_datetime(
    df["JoinDate"],
    errors="coerce"
)

print("JoinDate data type:", df["JoinDate"].dtype)
```

```python
# 6.6 HANDLE NUMERICAL MISSING VALUES
numeric_missing_columns = [
    "Age",
    "Salary",
    "PerformanceScore",
    "Satisfaction"
]

for column in numeric_missing_columns:

    df[column] = df[column].fillna(
        df[column].median()
    )
```

```python
# 6.7 HANDLE CATEGORICAL MISSING VALUES
df["Education"] = df["Education"].fillna(
    df["Education"].mode()[0]
)
```

7 . OutLier Detection

```python
# 13. OUTLIER DETECTION USING IQR

def detect_outliers(column):

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df
        [column] > upper_bound)
    ]

    print(f"Column: {column}")
    print(f"Q1: {Q1:.2f}")
    print(f"Q3: {Q3:.2f}")
    print(f"IQR: {IQR:.2f}")
    print(f"Lower Bound: {lower_bound:.2f}")
    print(f"Upper Bound: {upper_bound:.2f}")
    print(f"Number of Outliers: {len(outliers)}")

    return outliers
```

# 20. Key Business Insights

Based on the exploratory data analysis:

1. The department with the highest average salary was IT.

2. The department with the highest average performance score was IT.

3. Years of experience showed a strong positive relationship with salary.

4. Employees who worked overtime had a higher churn rate compared with employees who did not work overtime.

5. Employees who churned had an average satisfaction score of 3.10, compared with 3.32 for employees who did not churn.

6. The department with the highest churn rate was IT.

7. The relationship between performance score and salary was almost zero (very weak negative correlation).

8. The strongest numerical correlation identified was between Years of Experience and Salary.

# 21. Recommendations

Based on the findings from the analysis:

1. Departments with relatively high employee churn should be investigated
   to identify possible employee retention issues.

2. If overtime is associated with higher churn, management should monitor
   employee workload and overtime levels.

3. If lower satisfaction is associated with employee churn, management should
   consider employee satisfaction improvement initiatives.

4. Salary differences across departments and job levels should be reviewed
   to ensure compensation is appropriate for responsibilities and experience.
